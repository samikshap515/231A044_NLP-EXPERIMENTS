```python
print("Samiksha Patil", "UIN: 231A044\n")
import nltk
from nltk.corpus import wordnet as wn

# Ensure necessary NLTK resources are downloaded
nltk.download('wordnet')
nltk.download('punkt')

def lesk(word, sentence):
    """Implementation of the Lesk algorithm."""
    best_sense = None
    max_overlap = 0

    # Tokenize the sentence
    sentence_tokens = set(nltk.word_tokenize(sentence))

    # Get all senses for the word
    for sense in wn.synsets(word):
        # Get the definition (gloss) and examples of the sense
        definition = set(nltk.word_tokenize(sense.definition()))
        examples = set(nltk.word_tokenize(' '.join(sense.examples())))

        # Calculate overlap with the sentence
        overlap = len(definition.intersection(sentence_tokens)) + len(examples.intersection(sentence_tokens))

        if overlap > max_overlap:
            max_overlap = overlap
            best_sense = sense

    return best_sense
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
from nltk import word_tokenize
import nltk
nltk.download('punkt_tab')

d = "i am going to bank"
s = "i work in a bank"

example = "i like to go to bank"

len(
    list(set(word_tokenize(d)).intersection(word_tokenize(example)))
    + list(set(word_tokenize(s)).intersection(word_tokenize(example)))
)
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
sentence = "i book a ticket for my movie"
ambiguous_word = "book"

sense = lesk(ambiguous_word, sentence)

print(f"The best sense for '{ambiguous_word}' in the given context is: {sense.name()} - {sense.definition()}")
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
# Example dataset with annotated senses
dataset = [
    ("I went to the bank to deposit some money.", "bank.n.01"),
    ("He has a bank of knowledge.", "bank.n.02")
]

correct_predictions = 0

for sentence, correct_sense in dataset:
    predicted_sense = lesk("bank", sentence)

    if predicted_sense and predicted_sense.name() == correct_sense:
        correct_predictions += 1

accuracy = correct_predictions / len(dataset)
print(f"Accuracy: {accuracy:.2f}")
```
