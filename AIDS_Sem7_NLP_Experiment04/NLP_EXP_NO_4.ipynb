```python
print("Samiksha Patil", "UIN: 231A044\n")

import nltk
from nltk.tokenize import word_tokenize
from nltk.stem import WordNetLemmatizer
import string
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
# Download resources if not already done
nltk.download('punkt')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
import string
print(string.punctuation)
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
# Input text
# Step 1: Preprocessing
text_lower = text.lower()
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
#Tokenization
nltk.download('punkt_tab') # Download missing resource
tokens = word_tokenize(text_lower)

print(tokens)
```

```python
print("Samiksha Patil", "UIN: 231A044\n")

text_clean = []
for word in tokens:
    if word not in string.punctuation:
        text_clean.append(word)

print(text_clean)
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
# Step 3: Identify roots using morphological approximation using Lemmatizer

lemmatizer = WordNetLemmatizer()
lemmas = [lemmatizer.lemmatize(tok, pos='v') for tok in text_clean]  # Lemmatize as verbs

print(lemmas)
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
lemmatizer.lemmatize("computer", pos='n')
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
[(token, lemma) for token, lemma in zip(tokens, lemmas)]
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
# Step 4: Analyze tokens manually for derivational vs inflectional forms
morphological_info = {}

for token, lemma in zip(tokens, lemmas):
    if token != lemma:
        # This indicates some morphological change
        if token.endswith("ing"):
            analysis = "Inflectional: Present participle/continuous form of '{}'".format(lemma)
        elif token.endswith("er") or token.endswith("or"):
            analysis = "Derivational: Agent noun form of '{}'".format(lemma)
        else:
            analysis = "Inflectional or irregular form of '{}'".format(lemma)
    else:
        analysis = "Base/root form"

    morphological_info[token] = analysis
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
morphological_info
```
