```python
import nltk
nltk.download('punkt_tab')
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
from nltk import word_tokenize

tokenized = word_tokenize(text_6)

print("Tokenized words:", tokenized)
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
import string

punctuation = list(string.punctuation)

print(punctuation)
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
from string import punctuation
print("With punctuation:", tokenized)
print()

text_punct = []

for word in tokenized:
    if word not in punctuation:
        text_punct.append(word)

print("Without punctuation:", text_punct)
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
import nltk

nltk.download("stopwords")

from nltk.corpus import stopwords

print(stopwords.words("english"))
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
text_clean2 = []

for word in text_punct:
    if word not in stopwords.words("english"):
        text_clean2.append(word)

print(text_clean2)
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

# Create a list of stemmed words
text_stemmed_list = [stemmer.stem(word) for word in text_clean2]

# Print the resulting list
print(text_stemmed_list)
```
