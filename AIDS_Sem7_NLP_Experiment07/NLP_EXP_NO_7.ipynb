```python
print("Samiksha Patil", "UIN: 231A044\n")
# Sample text
text = """Microsoft is looking at buying U.K. startup for $1 billion.
Bill Gates founded Google."""

# Download the missing resource
import nltk
nltk.download('punkt_tab')

# Tokenize the text
tokens = word_tokenize(text)
print(tokens)
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
# Download the missing resource for part-of-speech tagging
import nltk
nltk.download('averaged_perceptron_tagger_eng')

# Perform part-of-speech tagging
tagged_tokens = pos_tag(tokens)

print(tagged_tokens)
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
# Download the missing resource for Named Entity Recognition
import nltk
nltk.download('maxent_ne_chunker_tab')

# Perform Named Entity Recognition (NER)
named_entities = ne_chunk(tagged_tokens)

# Display the NER tree
print(named_entities)
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
# Perform named entity recognition
named_entities = ne_chunk(tagged_tokens)

# Display named entities
for subtree in named_entities:
    if isinstance(subtree, nltk.Tree):
        entity = " ".join([word for word, tag in subtree.leaves()])
        label = subtree.label()
        print(f"Entity: {entity}, Label: {label}")
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
import spacy

# Load the spaCy English model
nlp = spacy.load("en_core_web_sm")

# Sample text
text = "Microsoft is looking at buying U.K. startup for $1 billion. Bill Gates founded Google."

# Process the text
doc = nlp(text)
print(doc)
# Extract named entities
for ent in doc.ents:
    print(f"Entity: {ent.text}, Label: {ent.label_}")
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
import nltk
from nltk import pos_tag, word_tokenize
from nltk.chunk import RegexpParser
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
sentence = "The quick brown fox jumps over the lazy dog."
sentence1 = "Microsoft is looking at buying U.K. startup for $1 billion. Bill Gates founded Google."
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
tokens = word_tokenize(sentence1)
tagged = pos_tag(tokens)
print(tagged)
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
# Defining a chunk grammar

grammar = "NP: {<DT>?<JJ>*<NN>}"
```

```python
print("Samiksha Patil", "UIN: 231A044\n")
# Create a Chunk Parser and Parse the Sentence
chunk_parser = RegexpParser(grammar)
chunked = chunk_parser.parse(tagged)

# Display the chunked output
print(chunked)
```
